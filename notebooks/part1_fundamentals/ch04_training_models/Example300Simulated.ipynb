```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
```

```python
# Correlation matrix
corr = np.array([
    [ 1.000,  0.970, -0.998, -0.960],
    [ 0.970,  1.000, -0.980, -0.990],
    [-0.998, -0.980,  1.000,  0.970],
    [-0.960, -0.990,  0.970,  1.000]
])

# Example labels
examples = {
    "Body": ["Weight", "Power", "Height", "Age"],
    "Vehicle": ["Weight", "Power", "Economy", "Engine"],
    "Survey": ["Item 1", "Item 2", "Item 3", "Item 4"]
}

# Choose the example to display
example = "Vehicle"
labels = examples[example]

# All unique off-diagonal pairs
pairs = [
    (0, 1), (0, 2), (0, 3),
    (1, 2), (1, 3), (2, 3)
]

# Create figure
fig, ax = plt.subplots(figsize=(9, 7))

im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)

ax.set_xticks(range(4))
ax.set_yticks(range(4))
ax.set_xticklabels(labels)
ax.set_yticklabels(labels)

ax.set_xlabel("Variables")
ax.set_ylabel("Variables")
ax.set_title("Correlation Matrix")

# Add colorbar only once
fig.colorbar(im, ax=ax, label="Correlation (r)")

# Display correlation values
for i in range(4):
    for j in range(4):
        ax.text(
            j, i, f"{corr[i, j]:+.3f}",
            ha="center", va="center", fontsize=11
        )

# Highlight the selected cell
highlight = plt.Rectangle(
    (0.5, -0.5), 1, 1,
    fill=False, edgecolor="yellow", linewidth=4
)
ax.add_patch(highlight)

# Information displayed below the matrix
info = fig.text(
    0.5, 0.02, "",
    ha="center", fontsize=12
)

def animate(frame):
    i, j = pairs[frame % len(pairs)]

    # Move the yellow highlight
    highlight.set_xy((j - 0.5, i - 0.5))

    # Describe the selected relationship
    value = corr[i, j]

    if abs(value) >= 0.7:
        strength = "Strong"
    elif abs(value) >= 0.4:
        strength = "Moderate"
    else:
        strength = "Weak"

    direction = "positive" if value > 0 else "negative"

    info.set_text(
        f"{labels[i]} and {labels[j]}: "
        f"{strength} {direction} correlation ({value:+.3f})"
    )

    return highlight, info

ani = FuncAnimation(
    fig,
    animate,
    frames=len(pairs),
    interval=1500,
    repeat=True,
    blit=False
)

plt.tight_layout(rect=[0, 0.07, 1, 1])

# Display the animation in Jupyter Notebook
plt.close(fig)
display(HTML(ani.to_jshtml()))
```
